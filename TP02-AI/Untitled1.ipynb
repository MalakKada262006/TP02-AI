```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
phrases = [
    "Le football est un sport passionnant et populaire.",
    "Le cinéma propose un film très captivant ce soir.",
    "Ce sport et ce film sont vraiment formidables.",
]

vectorizer_bow = CountVectorizer()
matrice_bow = vectorizer_bow.fit_transform(phrases)
vocabulaire = vectorizer_bow.get_feature_names_out()

df_bow = pd.DataFrame(matrice_bow.toarray(), columns=vocabulaire)

print("--- Liste des mots extraits (Vocabulaire) ---")
print(list(vocabulaire))
print("\n--- Matrice Bag of Words ---")
print(df_bow)
```
